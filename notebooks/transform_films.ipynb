# Exploratory Data Analysis of Films

```python
import pandas as pd
```

```python
# Create a new DataFrame for films
films_df = pd.read_csv("../data/raw/unclean-films.csv")

# Display info about the dataframe
films_df.info()
```

All the columns have the same number of values.

```python
# Check for null values
films_df.isnull().any(axis=1).sum()
```

There are no rows with missing values in any of the columns.

```python
# Check for any duplicates in the DataFrame
duplicates = films_df.duplicated().sum()
print(f"There are {duplicates} duplicate rows in the films DataFrame")
```

There are no duplicate rows in the films DataFrame so no need for cleaning.

```python
# Show the different date formats 
unique_date_formats = films_df['last_update'].unique()
print(f'Different date formats in last_update: {unique_date_formats}') 
```

```python
# Show the different data types in the release_year column
release_year_types = films_df['release_year'].map(type).unique()
print(f'Different data types in release_year column: {release_year_types}')
```
