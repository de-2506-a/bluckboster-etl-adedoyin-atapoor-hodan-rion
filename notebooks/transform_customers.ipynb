# Exploratory Data Analysis of Customers

```python
import pandas as pd
```

```python
# Create a new DataFrame for customers
customers_df = pd.read_csv("../data/raw/unclean-customers.csv")

# Display info about the dataframe
customers_df.info()
```

There are **3** missing data in the `address_id` column. 

```python
# Check for null values
customers_df.isnull().any(axis=1).sum()
```

There are **3** rows with missing values. This is **0.5%** of the total number of rows in the table so they can be safely removed. 

```python
# Check for any duplicates in the DataFrame
duplicates = customers_df.duplicated().sum()
print(f"There are {duplicates} duplicate rows in the customers DataFrame")
```

There are no duplicate rows in the films DataFrame so no need for cleaning.

```python
# Show the different date formats 
unique_date_formats = customers_df['create_date'].unique()
print(f'Different date formats in last_update: {unique_date_formats}') 
```

```python
# Show the different data types in the release_year column
release_year_types = customers_df['first_name'].map(type).unique()
print(f'Different data types in release_year column: {release_year_types}')
```
