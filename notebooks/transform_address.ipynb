# Exploratory Data Analysis of Address

```python
import pandas as pd
```

```python
# Create a new DataFrame for films
address_df= pd.read_csv("../data/raw/unclean-address.csv")

# Display info about the dataframe
address_df.info()
```

There are some missing values in `district`, `postal_code` and `phone` columns. 

```python
# Check for null values
address_df.isnull().any(axis=1).sum()
```

There are 603 rows with null values. This is because of the `address2` column which has no values. This can be safely dropped from the table as no data would be lost.

```python
# Check for any duplicates in the DataFrame
duplicates = address_df.duplicated().sum()
print(f"There are {duplicates} duplicate rows in the address DataFrame")
```

There are no duplicate roles in the address DataFrame so no need for cleaning.

```python
# Show the different date formats 
unique_date_formats = address_df['last_update'].unique()
print(f'Different date formats in last_update: {unique_date_formats}') 
```

```python
# Show the different data types in the release_year column
release_year_types = address_df['city_id'].map(type).unique()
print(f'Different data types in city_id column: {release_year_types}')
```
